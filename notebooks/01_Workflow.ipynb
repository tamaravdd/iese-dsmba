# Workflow with Python and AI

## Setting up Python

To write and run Python code, you need two things: a Python installation and an interface to work in. You have several options, from fully local setups to cloud-based environments:

**Option 1: Anaconda (easiest local setup)**. Download and install from `anaconda.com`. Anaconda comes pre-packaged with Python and most data science libraries used in this course. After installing, open **Anaconda Navigator** and start **Jupyter Notebook** from there. You're ready to go. For those who prefer the command line, you can bypass the Navigator and launch tools directly from the Terminal (Mac) or Anaconda Prompt (Windows).

**Option 2: uv (lightweight, modern)**. If you're comfortable with the command line, **uv** (`docs.astral.sh/uv`) is a fast, modern tool for managing Python projects. To get started:

1. Install uv (once): `curl -LsSf https://astral.sh/uv/install.sh | sh` (Mac/Linux) or `powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"` (Windows)
2. Create a project environment: `uv init myproject && cd myproject`
3. Add packages: `uv add numpy pandas matplotlib scikit-learn jupyter`
4. Launch Jupyter: `uv run jupyter notebook`

**Option 3: Google Colab (no installation)**. Go to [colab.google](https://colab.google/) with a Google account. Colab is a cloud-hosted notebook environment — it runs on Google's servers, so you don't need to install anything. Most packages are pre-installed. However, Colab has limitations (e.g. file persistence, custom package versions, some libraries won't work) so it won't be suitable for everything in this course.

### Jupyter notebooks

Whichever option you choose, you will be working with **Jupyter notebooks**. The Jupyter ecosystem introduces the **notebook** paradigm, allowing seamless integration of code, output, and documentation in a single document. The page that you are reading right now is a rendered notebook.

Historically, developers used **IPython** (Interactive Python), which enhanced the basic Python shell with extra features. IPython introduced **magic commands**, denoted by `%cmd` (like `%cd` for changing directories). While these commands aren't part of Python itself, they're still referenced in many tutorials. For a complete list, use `%quickref`.

### IDEs

Beyond notebooks, you may encounter full **Integrated Development Environments (IDEs)** like **Visual Studio Code** or **Cursor**. These provide comprehensive development tools like consoles, text editors, debuggers, and variable inspectors. Feel free to use any tool you're comfortable with, though our instructions will assume you're using Jupyter Notebook or Jupyter Lab. The concepts transfer to other interfaces with minimal adaptation.
